```python
import os
import pandas as pd

root = r"./"  # 根目录（可改成你的路径）

data = []

for folder in os.listdir(root):
    folder_path = os.path.join(root, folder)
    if not os.path.isdir(folder_path):
        continue

    csv_path = os.path.join(folder_path, "labeled_data.csv")
    if not os.path.exists(csv_path):
        continue

    # ---- 根据命名规则解析 ----
    # 示例: "0204-7ST-3L2"
    parts = folder.split('-')
    if len(parts) < 3:
        continue

    gesture = parts[1][1:3]   # "ST"、"G"、"P"
    pose = parts[2][0]        # "1"、"2"、"3"
    direction = parts[2][1]   # "B"、"L"、"R"、"U"

    # ---- 统计 ----
    size_mb = os.path.getsize(csv_path) / (1024 * 1024)
    with open(csv_path, 'r', encoding='utf-8', errors='ignore') as f:
        rows = sum(1 for _ in f) - 1  # 去掉header

    data.append([gesture, pose, direction, 1, size_mb, rows])

# ---- 聚合统计 ----
df = pd.DataFrame(data, columns=['gesture', 'pose', 'direction', 'file_count', 'size_mb', 'rows'])
summary = df.groupby(['gesture', 'pose', 'direction']).agg({
    'file_count': 'sum',
    'size_mb': 'sum',
    'rows': 'sum'
}).reset_index()

# ---- 输出 ----
summary['size_mb'] = summary['size_mb'].round(3)
print(summary)
summary.to_csv("summary.csv", index=False)
print("\n✅ 已保存到 summary.csv")
```
